```python
'''
Ini adalah code untuk menyambungkan notebook dengan server bigquery
'''
from google.colab import auth
from google.cloud import bigquery
auth.authenticate_user()
print('Authenticated')

project_id_akun = "dummy-456310" # Gunakan GCP Project-ID Anda
client = bigquery.Client(project=project_id_akun)
```

```python
'''
Ini adalah code untuk pengambilan data dari bigquery
'''
df_1 = client.query(
    '''
SELECT *
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`

'''
).to_dataframe()

df_1 # menampilkan dataframe
```

```python
'''
Ini adalah code untuk menampilkan unique value dari kolom company

penjelasan:
kita menampilkan unique value dengan menggunakan distinct dari kolom company

analisa:
Dari sini kita bisa melihat bahwa ada 5 company yang berbeda dalam penyediaan layanan taksi yang terdapat pada dataset tersebut.
company tersebut yaitu: Atomic Inc, Deluc Porter,Federway Int, Synctaxi Inc, dan Rocket Cab Auto.

'''
df_1 = client.query(
    '''
SELECT distinct company
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`

'''
).to_dataframe()

df_1 # menampilkan dataframe

```

```python
'''
Ini adalah code untuk memperlihatkan data minimal dan data maximal dari kolom trip_start_timestamp

penjelasan:
kita menggunakan fungsi min dari kolom trip_start_timestamp lalu diberi nama sebagai min_pengambilan data
lalu menggunakan fungsi max dari kolom trip_start_timestamp lalu diberi nama sebagai max_pengambilan_data

analisa:
DAri sini kita bisa melihat bahwa untuk minimal waktu pengambilan data adalah pada tanggal 2 february 2021 pukul 11 pagi (diwakilkan oleh kolom min_pengambilan_data)
dan waktu maksimum dari pengambilan data yaitu pada tanggal 31 december 2022 pukul 14.15 (diwakilkan oleh kolom max_pengambilan_data)

Saya menggunakan trip_start_timestamp dan bukan trip_end_timestamp karena data dibuat pada saat pelanggan memesan transaksi bukan pada saat mereka selesai transaksi
'''
df_1 = client.query(
    '''
SELECT min(trip_start_timestamp) as min_pengambilan_data, max(trip_start_timestamp) as max_pengambilan_data
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`

'''
).to_dataframe()

df_1
```

```python
'''
Ini adalah code untuk memperlihatkan dari seluruh total perjalanan (km) untuk setiap company pada tahun 2022

penjelasan:
kita menghitung total trip dengan menjumlahkan kolom trip_km berdasarkan company masing masing lalu dibebri nama total perjalanan

analisa:
Disini kita melihat bahwa perjalanan yang ditempuh terbanyak yaitu company Deluc Porter	dengan 18515.21 km
Dan yang terkecil adalah company Rocket Cab Auto sebanyak 1031.09 km
Disini hanya tercantum 4 company karena 1 company yang lain yaitu Atomic Inc. tidak melakukan transaksi pada tahun 2022
'''
df_1 = client.query(
    '''
SELECT company, sum(trip_km) as total_perjalanan
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`
where trip_start_timestamp >= '2022-01-01' and trip_start_timestamp <= '2022-12-31'
group by company


'''
).to_dataframe()

df_1
```

```python
'''
Ini code untuk mencari average price per km berdasarkan company

penjelasan:
disini kita menggunakan subquery.

langkah pertama yaitu kita mencari nilai average dari jarak berdasarkan company nya lalu disimpan sebagai avg_trip_km
lalu mencari nilai average dari harga berdasarkan company nya juga dan disimpan sebagai avg_fare_company
setelah menjadi 1 query yang memiliki nilai avg_trip_km dan avg_fare_company
kita mengambil mencari nilai avg_price_per_km dengan menggunakan rumus avg_fare_company dibagi dengan avg_trip_km

analisa:
Dari sini kita bisa melihat bahwa company Atomic Inc.	memiliki harga paling mahal yaitu dengan tarif 2.40 per km nya
lalu untuk yang paling murah adalah company Rocket Cab Auto	dengan taruk 1.67 per km nya
'''
df_1 = client.query(
    '''

select x.company, x.avg_fare_company/avg_trip_km as avg_price_per_km from (
SELECT company, avg(trip_km) as avg_trip_km, avg(fare) as avg_fare_company
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`
group by company
) as x


'''
).to_dataframe()

df_1
```

**2. Pandas**

```python
'''
pembuatan dataframe menggunakan pandas
'''
df_1 = client.query(
    '''

SELECT *
FROM `ftds-hacktiv8-project.phase0_ftds_027_hck.taxi_fare`


'''
).to_dataframe()

df_1 # menampilkan dataframe
```

```python
df_1['period']= df_1['trip_start_timestamp'].dt.hour
df_1['period'] = df_1['period'].map(str)
df_1['period']
```

```python
import numpy as np

df_1['period'] = np.where(df_1['period'] >= 3 & df_1['period'] <= 10, 'Morning', df_1['period'])
df_1
```

```python
'''
Code ini untuk menampilkan data dari perusahaan Federway Int. pada tahun 2022.

penjelasan:
saya membuat dataframe baru yang bernama df_baru untuk menampung data yang akan dicari.
Data tersebut didapat dari pengambilan data pada df_1 dengan kolom company yang bernilai 'Federway Int.' dan harus memiliki rentan tanggal 1 january 2022 hingga 31 desember 2022 pada kolom trip_end_timestamp
nanti dataframe tersebut akan diolah lagi untuk mencari tahu total jarak perjalanan di bawah codingan ini.

analisa:
Dari data ini kita dapat menyimpulkan bahwa ada 1059 transaksi dari perusahaan Federway Int. pada tahun 2022
'''

df_baru = df_1[(df_1['company'] == 'Federway Int.') & ((df_1['trip_end_timestamp'] >= '2022-01-01') & (df_1['trip_end_timestamp'] <= '2022-12-31'))]
df_baru
```

```python
'''
Code ini untuk menampilkan total perjalanan yang dicapai perusahaan Federway Int. pada tahun 2022

penjelasan:
dari hasil dataframe yang sebelum ini yaitu df_baru. kita mengambil total trip_km nya menggunakan metode .sum()
caranya adalah setiap nilai yang ada di kolom trip_km pada df_baru akan di tambahkan semuanya dan dicantumkan hasil penjumlahannya di paling bawah

analisa:
dari 1060 transaksi yang dilakukan oleh company Federway Int. pada tahun 2022, Mereka telah menyelesaikan trip sejauh 4359.24 km.
'''

import pandas as pd

df_baru.loc['trip_km'] = pd.Series(df_baru['trip_km'].sum(), index=['trip_km'])
df_baru
```
